```python

import matplotlib.pyplot as plt
pos_gap = test_object.get_pos_gap_for_joints_TEST()

plt.figure(figsize=(20, 5))
plt.title("gap of the position, velocity, acceleration and their sum averaged over all frames")
plt.xlabel("joint")
plt.ylabel("pos/vel/acc/total gap")
plt.plot([JOINT_SHORT[joint_name] for joint_name in pos_gap.keys()], [x for x,y in pos_gap.values()], "o", color="red", label="avg")
plt.plot([JOINT_SHORT[joint_name] for joint_name in pos_gap.keys()], [x+y for x,y in pos_gap.values()], "o", color="yellow", label="var")
plt.plot([JOINT_SHORT[joint_name] for joint_name in pos_gap.keys()], [x-y for x,y in pos_gap.values()], "o", color="yellow", label="var")
plt.legend()
plt.show()
```
