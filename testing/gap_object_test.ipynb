```python
from src import *
import matplotlib.pyplot as plt

logger = logging.getLogger("global_logger")
logging.basicConfig(filename='../gap_object_test.log', format='%(levelname)s: %(message)s', encoding='utf-8', filemode='w', level=logging.DEBUG)
logger.info("Starting gap_object_test.")

replay_settings = SimulationParameters("default")
simulator = SimulatorHandler()
data = [(action_name, None, None) for action_name in ACTION_NAMES]
gap_handler = simulator.simulation_gap(settings=replay_settings, data=data)

# define which logs to use for testing
action_name = "standup_front"
recording_date = "260108"
log_index = 4
joint_name = "lAnklePitch"
start_index = 0
end_index = -1

gap_handler_visualizer = GapHandlerVisualizer(gap_handler)
gap_object_visualizer = gap_handler_visualizer.get_gap_object_visualizer(action_name, recording_date, log_index)

test_object = gap_handler.get(action_name, recording_date, log_index)

print(test_object.get_pos_gap_avg())
print(test_object.get_vel_gap_avg())
print(test_object.get_acc_gap_avg())
print(test_object.get_total_gap_avg())
```

```python
x_axis = test_object.get_time_steps(start_index, end_index)
extraction = test_object.get_gyro_pos_extraction(start_index=start_index, end_index=end_index)
replay = test_object.get_gyro_pos_replay(start_index=start_index, end_index=end_index)
y_label = "Position (degrees)/Velocity (?)/Acceleration (?)"

plt.figure(figsize=(20, 5))
plt.title("Gyro Orientation")
plt.xlabel("time step")
plt.ylabel("gyro orientation in degrees")
plt.grid(visible=True)
plt.plot(x_axis, extraction[0], linestyle="-", color="mistyrose", label="e_x", marker=".")
plt.plot(x_axis, extraction[1], linestyle="-", color="red", label="e_y", marker=".")
plt.plot(x_axis, extraction[2], linestyle="-", color="darkred", label="e_z", marker=".")
plt.plot(x_axis, replay[0], linestyle="-", label="r_x", color="lightblue", marker=".")
plt.plot(x_axis, replay[1], linestyle="-", label="r_y", color="blue", marker=".")
plt.plot(x_axis, replay[2], linestyle="-", label="r_z", color="darkblue", marker=".")
plt.legend()
plt.show()

extraction = test_object.get_gyro_vel_extraction(start_index=start_index, end_index=end_index)
replay = test_object.get_gyro_vel_replay(start_index=start_index, end_index=end_index)

plt.figure(figsize=(20, 5))
plt.title("Gyro Velocity")
plt.xlabel("time step")
plt.ylabel("gyro velocity in deg/s")
plt.grid(visible=True)
plt.plot(x_axis, extraction[0], linestyle="-", color="mistyrose", label="e_x", marker=".")
plt.plot(x_axis, extraction[1], linestyle="-", color="red", label="e_y", marker=".")
plt.plot(x_axis, extraction[2], linestyle="-", color="darkred", label="e_z", marker=".")
plt.plot(x_axis, replay[0], linestyle="-", label="r_x", color="lightblue", marker=".")
plt.plot(x_axis, replay[1], linestyle="-", label="r_y", color="blue", marker=".")
plt.plot(x_axis, replay[2], linestyle="-", label="r_z", color="darkblue", marker=".")
plt.legend()
plt.show()

extraction = test_object.get_gyro_acc_extraction(start_index=start_index, end_index=end_index)
replay = test_object.get_gyro_acc_replay(start_index=start_index, end_index=end_index)

plt.figure(figsize=(20, 5))
plt.title("Gyro Acceleration")
plt.xlabel("time step")
plt.ylabel("gyro acceleration in deg/s^2")
plt.grid(visible=True)
plt.plot(x_axis, extraction[0], linestyle="-", color="mistyrose", label="e_x", marker=".")
plt.plot(x_axis, extraction[1], linestyle="-", color="red", label="e_y", marker=".")
plt.plot(x_axis, extraction[2], linestyle="-", color="darkred", label="e_z", marker=".")
plt.plot(x_axis, replay[0], linestyle="-", label="r_x", color="lightblue", marker=".")
plt.plot(x_axis, replay[1], linestyle="-", label="r_y", color="blue", marker=".")
plt.plot(x_axis, replay[2], linestyle="-", label="r_z", color="darkblue", marker=".")
plt.legend()
plt.show()
```
